```python
# prepare data

marks = [[98],[80],[40],[20],[85],[10],[30]]  # input
result = ['Pass','Pass','Fail','Fail','Pass','Fail','Fail'] # output
```

```python
['Pass' if val[0]>50 else "Fail" for val in marks]
```

```python
# For all ML operations we have one of the biggest Library
# Scikit-learn
# But we import it as sklearn
```

```python
# our problem is Classification problem
# import algorithm
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model
```

```python
# Train the model
model.fit(marks,result)  # input,output
```

```python
model.classes_
```

```python
model.coef_
```

```python
# Test the model prediction
marks,result
```

```python
# TESTING
model.predict([[51]])  # 50% =< prob,then its FAIL else PASS
```

```python
# to check the probability

model.predict_proba([[60]])
```

```python
model.predict_proba([[60],[58],[90],[66]])
```
